```python
import os
import tempfile
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import scanpy as sc
import scvi
import seaborn as sns
from scvi.external import MRVI
from plotnine import *
import xarray as xr
from PyComplexHeatmap import ClusterMapPlotter, HeatmapAnnotation, anno_simple, anno_label
```

```python
# ==============================================================================
# 1. LOAD DATA & MODELS
# ==============================================================================

# Load AnnData object
adata = sc.read_h5ad("/mnt/data/melina/adata_all_batch_Epithelial.h5ad")

# Load MRVI Model (final mrvi model from job 171, sample_key = sample_ID, batch_key = donor)
model_mrvi = MRVI.load("/mnt/data/melina/models/mrvi_model_Epithelial", adata=adata)

# Load DE results dataset
dmats = xr.open_dataset("/mnt/data/melina/deg_Epithelial.nc")
ds = dmats
de_res = dmats

# Inspect the structure (dimensions, coordinates, and variables)
print(ds)
```

```python
# ==============================================================================
# 2. MAP EFFECT SIZE & PLOT UMAP (WITHOUT LABELS)
# ==============================================================================

# Define your group
comp_group = ["donor_disease_category_UC_disease", "donor_disease_category_UC_control"]

for i in comp_group:
    # Map the effect size to observations (Ensure alignment matches adata.obs_names)
    adata.obs[i] = de_res.effect_size.sel(covariate=i).values

    # Calculate a robust max for the color scale of the current effect size
    vmax_val = float(np.nanquantile(adata.obs[i], 0.95))

    # Plot UMAP embedding with mapped effect sizes (labels and colorbars removed)
    sc.pl.embedding(
        adata,
        basis="umap_mrvi_u",
        color=[i],
        ncols=2,                   
        layer="data",              
        cmap="viridis", 
        vmax=vmax_val,             
        frameon=False,
        title="",                  # Removes the plot title
        colorbar_loc=None          # Removes the colorbar scale and text
    )


# ==============================================================================
# 3. MITOCHONDRIAL PDE VIOLIN PLOT (WITHOUT LABELS)
# ==============================================================================

# Automatically identify mitochondrial genes present in your specific dataset
mt_genes = [g for g in adata.var_names if g.upper().startswith('MT-')]

# Loop over each target covariate to generate clean violin plots sequentially
for target_cov in comp_group:
    # Slice the dataset for MT genes AND the specific covariate
    mt_pde_uc = ds['pde'].sel(gene=mt_genes, covariate=target_cov)

    # Convert to a standard DataFrame for Seaborn plotting
    mt_pde_df_uc = mt_pde_uc.to_dataframe(name='pde_value').reset_index()

    # Create the Plot
    plt.figure(figsize=(14, 6))
    sns.violinplot(
        data=mt_pde_df_uc, 
        x='gene', 
        y='pde_value', 
        inner='quartile', 
        palette='magma'
    )

    # Stripping all labels, text, titles, and ticks
    plt.title('')
    plt.xlabel('')
    plt.ylabel('')
    plt.xticks([])
    plt.yticks([])

    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()
```

```python
# ==============================================================================
# 2. MAP EFFECT SIZE & PLOT UMAP (WITH LABELS)
# ==============================================================================

# Define your group
comp_group = ["donor_disease_category_UC_disease", "donor_disease_category_UC_control"]

for i in comp_group:
    # Map the effect size to observations (Ensure alignment matches adata.obs_names)
    adata.obs[i] = de_res.effect_size.sel(covariate=i).values

    # Calculate a robust max for the color scale of the current effect size
    vmax_val = float(np.nanquantile(adata.obs[i], 0.95))

    # Plot UMAP embedding with mapped effect sizes
    sc.pl.embedding(
        adata,
        basis="umap_mrvi_u",
        color=[i],
        ncols=2,                   
        layer="data",              
        cmap="viridis", 
        vmax=vmax_val,             
        frameon=False,
        title=f"Effect Size: {i}",   # Dynamically adds the title back
        colorbar_loc="right"         # Restores the colorbar on the right side
    )


# ==============================================================================
# 3. MITOCHONDRIAL PDE VIOLIN PLOT (WITH LABELS)
# ==============================================================================

# Automatically identify mitochondrial genes present in your specific dataset
mt_genes = [g for g in adata.var_names if g.upper().startswith('MT-')]

# Loop over each target covariate to generate comprehensive violin plots sequentially
for target_cov in comp_group:
    # Slice the dataset for MT genes AND the specific covariate
    mt_pde_uc = ds['pde'].sel(gene=mt_genes, covariate=target_cov)

    # Convert to a standard DataFrame for Seaborn plotting
    mt_pde_df_uc = mt_pde_uc.to_dataframe(name='pde_value').reset_index()

    # Create the Plot
    plt.figure(figsize=(14, 6))
    sns.violinplot(
        data=mt_pde_df_uc, 
        x='gene', 
        y='pde_value', 
        inner='quartile', 
        palette='magma'
    )

    # Restoring all labels, text, titles, and ticks
    plt.title(f'Mitochondrial Probability of DE (PDE) in {target_cov}', fontsize=16)
    plt.ylabel("Probability of DE", fontsize=12)
    plt.xlabel("Mitochondrial Gene", fontsize=12)
    
    # Restoring tick labels and adjusting x-axis rotation for clarity
    plt.xticks(rotation=45, ha='right')
    plt.yticks() 

    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()
```
